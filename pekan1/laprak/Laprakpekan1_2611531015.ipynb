## 2.1 Dasar Teori
Konsep dasar dan alat yang digunakan dalam praktikum ini meliputi beberapa aspek utama:

1. **Python dan Fungsi `print()`**: Python adalah bahasa pemrograman tingkat tinggi berjenis *interpreted*. Fungsi `print()` dalam Python digunakan untuk menampilkan teks atau nilai variabel ke konsol/layar. Pemrosesan string pada fungsi `print()` mendukung penggabungan (*string concatenation*) menggunakan operator `+` serta karakter khusus (*escape character*) seperti `\\n` (baris baru) dan `\\\\` (mencetak karakter *backslash*).
2. **Komentar (*Comment*)**: Komentar dalam Python digunakan untuk memberikan dokumentasi atau penjelasan pada kode tanpa dieksekusi oleh interpreter. Komentar baris tunggal diawali simbol `#`, sedangkan komentar *multi-line* umumnya menggunakan *triple quotes* (`\"\"\"` atau `'''`).
3. **Visual Studio Code (VSCode)**: VSCode adalah *text editor* ringan namun *powerful* yang dikembangkan oleh Microsoft. VSCode mendukung berbagai ekstensi seperti *Python Extension* dan *Jupyter Notebook* yang mempermudah eksekusi instruksi program.
4. **Git dan GitHub**: Git adalah *Version Control System* (VCS) terdistribusi yang mencatat setiap perubahan pada berkas kode. GitHub adalah platform berbasis *cloud* tempat menyimpan dan berbagi repositori Git, memungkinkan pencadangan kode dan kolaborasi terstruktur.

## 2.2 Kode Program

### 2.2.1 Program "A_2611531015"
Jalankan sel kode di bawah ini agar Jupyter otomatis menyimpan kodenya beserta outputnya sebagai satu kesatuan (tidak perlu screenshot terpisah untuk hasil run).

```python
# Program pertama
"""
Komentar
dua baris atau lebih
pertemuan 1 praktikum alpro
"""
print("Hello World")
print()
print("Program ini menghasilkan")
print("Empat baris output")
print("/ \\ // \\\\ /// \\\\\\")
print("program ini " + "dibuat oleh a/n " + "Ahmad Affandi dan 2611531015")
```

**Penjelasan :**
1. **Bagian Komentar**: Simbol `#` digunakan untuk komentar satu baris, sedangkan tanda petik tiga (`\"\"\"`) digunakan untuk komentar *multi-line*. Seluruh blok teks komentar ini diabaikan oleh *interpreter* Python sehingga tidak memunculkan output.
2. **Fungsi `print()` Dasar dan Baris Kosong**: `print(\"Hello World\")` mencetak teks ke layar, diikuti oleh `print()` kosong yang berfungsi memberikan satu baris jarak (*newline*).
3. **Penggunaan Karakter *Escape* (`\\\\`)**: Pada baris `print(\"/ \\\\ // \\\\\\\\ /// \\\\\\\\\\\\\")`, karakter *backslash* (`\\`) bertindak sebagai karakter *escape*. Oleh karena itu, penulisan dua pasang *backslash* (`\\\\`) akan dieksekusi dan ditampilkan sebagai satu *backslash* tunggal (`\\`) pada terminal output.
4. **Penggabungan String (*String Concatenation*)**: Pada baris terakhir, tiga potongan string digabungkan menjadi satu kalimat utuh menggunakan operator `+` yang menampilkan identitas pembuat program.

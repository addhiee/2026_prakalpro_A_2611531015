## 1.1 Latar Belakang
Dalam pemrosesan program komputer, operator merupakan simbol-simbol khusus yang digunakan untuk melakukan operasi tertentu terhadap satu atau beberapa nilai (operan). Memahami cara kerja operator merupakan fondasi utama dalam membangun alur logika program, mulai dari perhitungan matematis sederhana hingga pengondisian alur logika yang kompleks.

Bahasa pemrograman Python menyediakan berbagai jenis operator yang sangat fleksibel. Selain operator aritmatika dan perbandingan standar, Python mendukung operator keanggotaan (`in`, `not in`) untuk memeriksa elemen dalam koleksi data, operator identitas (`is`, `is not`) untuk mengevaluasi alamat memori objek, serta operator bitwise untuk melakukan pemrosesan data langsung pada tingkat bit biner. Pemahaman mengenai perbedaan perilaku setiap operator, seperti perbedaan perbandingan nilai (`==`) dan identitas (`is`), sangat krusial untuk menghindari bug atau kesalahan logika saat membangun aplikasi.

## 2.1 Dasar Teori
Konsep-konsep utama mengenai tipe data dan variabel pada bahasa Python meliputi:

1. **Variabel dan Dynamically Typed**: Variabel adalah pemetaan nama yang menunjuk ke lokasi memori penampung nilai. Dalam Python, variabel tidak perlu dideklarasikan tipe datanya terlebih dahulu. Interpreter akan mengevaluasi tipe data secara otomatis saat alokasi nilai dilakukan.
2. **Tipe Data Numerik**: Python mendukung tiga jenis numerik utama:
    - **Integer (`int`)**: Bilangan bulat positif atau negatif tanpa koma desimal.
    - **Float (`float`)**: Bilangan riil berpresisi presisi ganda yang mengandung titik desimal.
    - **Complex (`complex`)**: Bilangan kompleks berbentuk $a + bj$, di mana $a$ adalah bagian riil dan $b$ adalah bagian imajiner.
3. **Tipe Data String (`string`)**: Kumpulan karakter yang dibatasi oleh tanda petik tunggal (`'`), tanda petik ganda (`\"`), atau *triple quotes* (`\"\"\"` / `'''`). Tanda petik tiga memungkinkan penulisan teks multi-baris (*multiline string*) yang mempertahankan format spasi dan baris baru.
4. **Tipe Data Boolean (`bool`)**: Tipe data logika yang hanya memiliki dua kemungkinan nilai, yaitu `True` atau `False`. Nilai boolean sering kali dihasilkan dari evaluasi ekspresi kondisi atau operator perbandingan (misalnya `>=`, `<=`, `==`).
5. **Input dan Type Casting**: Fungsi `input()` digunakan untuk mengambil masukan dari pengguna secara interaktif melalui konsol. Karena keluaran bawaan fungsi `input()` selalu berbentuk String (`string`), maka diperlukan *type casting* menggunakan fungsi seperti `int()` atau `float()` agar nilai tersebut dapat dioperasikan secara matematis.
6. **Konstanta dan String Formatting**: Menurut standar konvensi PEP 8, konstanta dinyatakan dengan nama variabel berhuruf kapital. Dalam Python modern, kejelasan konstanta dipertegas dengan penanda `Final` dari modul `typing`. Untuk menampilkan nilai numerik atau konstanta secara rapi, digunakan *string formatting* (seperti operator `%` atau `%f` / `%.2f`).

### 2.2.1 Program "aritmatika_2611531015.py"
Jalankan sel kode di bawah ini agar Jupyter otomatis menyimpan kodenya beserta outputnya sebagai satu kesatuan (tidak perlu screenshot terpisah untuk hasil run).

```python
# Buat file dengan nama aritmatika_NIM.py
# Buatb program untuk operator artimatika dalam python
# Nama variabel ditambah 4 digit nim terakhir
# Program ini menggunakan fungsi input() 
#  Nilai yang dimasukkan akan dikonversi menjadi tipe data integer

angka1_1015 = int(input("Masukkan angka pertama: "))
angka2_1015 = int(input("Masukkan angka kedua: "))

#Penjumlahan
hasil_1015 = angka1_1015 + angka2_1015
print("\nOperator Penjumlahan")
print("Hasil =", hasil_1015)

#Pengurangan
hasil_1015 = angka1_1015 - angka2_1015
print("\nOperator Pengurangan")
print("Hasil =", hasil_1015)

#Perkalian
hasil_1015 = angka1_1015 * angka2_1015
print("\nOperator Perkalian")
print("Hasil =", hasil_1015)

#Pembagian, Pembagian bulat, dan sisa bagi
if angka2_1015 != 0:
    hasil_1015 = angka1_1015 / angka2_1015
    print("\nOperator Pembagian")
    print("Hasil =", hasil_1015)

    hasil_1015 = angka1_1015 // angka2_1015
    print("\nOperator Pembagian Bulat")
    print("Hasil =", hasil_1015)

    hasil_1015 = angka1_1015 % angka2_1015
    print("\nOperator Sisa Bagi")
    print("Hasil =", hasil_1015)

else:
    print("Angka kedua tidak boleh bernilai 0")

# Pangkat
hasil_1015 = angka1_1015 ** angka2_1015
print("\nOperator Pangkat")
print("Hasil =", hasil_1015)
```

**Penjelasan :**
1. **Input Data dan Type Casting**: Program menerima dua masukan angka dari pengguna menggunakan fungsi `input()`. Nilai masukan secara otomatis dikonversi dari string menjadi bilangan bulat (`int`) menggunakan *type casting* `int()`.
2. **Operasi Aritmatika Dasar**: Melakukan kalkulasi matematika dasar meliputi penjumlahan (`+`), pengurangan (`-`), dan perkalian (`*`) antara variabel `angka1_1015` dan `angka2_1015`.
3. **Pengondisian dan Operasi Pembagian**: Menggunakan struktur `if-else` untuk mengecek apakah `angka2_1015 != 0` guna menghindari kesalahan pembagian dengan nol (*division by zero*). Jika valid, program menjalankan pembagian riil (`/`), pembagian bulat (`//`), dan sisa bagi (`%`).
4. **Operasi Perpangkatan (`**`)**: Menghitung hasil pangkat dari `angka1_1015` dipangkatkan dengan `angka2_1015`.

### 2.2.2 Program "perbandingan_2611531015.py"
Jalankan sel kode di bawah ini agar Jupyter otomatis menyimpan kodenya beserta outputnya sebagai satu kesatuan (tidak perlu screenshot terpisah untuk hasil run).

```python
# Buat file dengan nama perbadingan_NIM.py
# Nama variabel ditambah 4 digit nim terakhir
# Program ini menggunakan fungsi input()
# Nilai yang dimasukkan akan dikonversi menjadi tipe data integer
# Program operator perbandingan dalam python

angka1_1015 = int(input("Masukkan angka pertama: "))
angka2_1015 = int(input("Masukkan angka kedua: "))

# Lebih besar dari
hasil_1015 = angka1_1015 > angka2_1015
print("\nOperator Lebih Besar Dari")
print("angka1 > angka2 =", hasil_1015)

# Lebih kecil dari
hasil_1015 = angka1_1015 < angka2_1015
print("\nOperator Lebih Kecil Dari")
print("angka1 < angka2 =", hasil_1015)

# Lebih besar dari atau sama dengan
hasil_1015 = angka1_1015 >= angka2_1015
print("\nOperator Lebih Besar Dari atau Sama Dengan")
print("angka1 >= angka2 =", hasil_1015)

# Lebih kecil dari atau sama dengan
hasil_1015 = angka1_1015 <= angka2_1015
print("\nOperator Lebih Kecil Dari atau Sama Dengan")
print("angka1 <= angka2 =", hasil_1015)

# Sama dengan
hasil_1015 = angka1_1015 == angka2_1015
print("\nOperator Sama Dengan")
print("angka1 == angka2 =", hasil_1015)

# Tidak sama dengan
hasil_1015 = angka1_1015 != angka2_1015
print("\nOperator Tidak Sama Dengan")
print("angka1 != angka2 =", hasil_1015)

# Tambahan: Perbadingan berantai dalam python
hasil_1015 = 0 < angka1_1015 < 100
print("\nOperator Perbandingan Berantai")
print("0 < angka1 < 100 =", hasil_1015)

hasil_1015 = 0 < angka2_1015 < 100
print("0 < angka2 < 100 =", hasil_1015)
```

**Penjelasan :**
1. **Input Data**: Mengambil dua input berupa nilai integer yang akan dibandingkan.
2. **Operator Perbandingan (Relasional)**: Menguji hubungan antara dua variabel angka yang menghasilkan keluaran bertipe boolean (`True` atau `False`), yaitu `>`, `<`, `>=`, `<=`, `==`, dan `!=`.
3. **Perbandingan Berantai (*Chained Comparison*)**: Mengevaluasi apakah suatu angka berada di dalam rentang tertentu secara sekaligus (contoh: `0 < angka1_1015 < 100`) tanpa perlu menggabungkan dua ekspresi terpisah.

### 2.2.3 Program "logika_2611531015.py"
Jalankan sel kode di bawah ini agar Jupyter otomatis menyimpan kodenya beserta outputnya sebagai satu kesatuan (tidak perlu screenshot terpisah untuk hasil run).

```python
# buat file dengan nama logika_NIM.py
# Nama variabel ditambah 4 digit nim terakhir
# Program ini menggunakan fungsi input()
# Program   operator logika dalam python

# Memasukkan Nilai boolean
# input tidak peka terhadap huruf besar dan kecil
a1_1015 = ("input nilai boolean-1 (True/False): "). strip().lower() == "true"
a2_1015 = ("input nilai boolean-2 (True/False): "). strip().lower() == "true"

print ("\nA1= ", a1_1015)
print ("A2= ", a2_1015)

# konjungsi: bernilai true jika keduanya true
hasil_1015 = a1_1015 and a2_1015
print("\nOperator Konjungsi (AND)")
print("A1 and A2 =", hasil_1015)

# Disjungsi: bernilai true jika salah satu true
hasil_1015 = a1_1015 or a2_1015
print("\nOperator Disjungsi (OR)")
print("A1 or A2 =", hasil_1015)

# Negasi A1: membalik nilai A1
hasil_1015 = not a1_1015
print("\nOperator Negasi (NOT)")
print("not A1 =", hasil_1015)

# Negasi A2: membalik nilai A2
hasil_1015 = not a2_1015
print("not A2 =", hasil_1015)

# XOR: bernilai True jika kedua nilai berbeda
hasil_1015 = a1_1015 != a2_1015
print("\nDijungsi Eksklusif (XOR)")
print("A1 XOR A2 =", hasil_1015)
```

2. **Operator Logika Dasar**: Evaluasi konjungsi (`and`), disjungsi (`or`), dan negasi (`not`).
3. **Disjungsi Eksklusif (XOR)**: Diimplementasikan menggunakan operator `!=` antara dua nilai boolean, yang bernilai `True` hanya jika kedua input memiliki nilai logika yang berbeda.

### 2.2.4 Program "assignment_2611531015.py"
Jalankan sel kode di bawah ini agar Jupyter otomatis menyimpan kodenya beserta outputnya sebagai satu kesatuan (tidak perlu screenshot terpisah untuk hasil run).

```python
# Buat file dengan nama assignment_NIM.py
# Nama variabel ditambah 4 digit nim terakhir
# Program ini menggunakan fungsi input()
# Nilai yang dimasukkan akan dikonversi menjadi tipe data integer
# Program operator assignment dalam python

angka1_1015 = int(input("Input Angka-1: "))
angka2_1015 = int(input("Input Angka-2: "))

print("\nNilai awal angka1 = ", angka1_1015)
print("Nilai angka2 = ", angka2_1015)

# assignment biasa
hasil_1015 = angka1_1015
print("Assignment Biasa (=)")
print("Hasil =", hasil_1015)

# assignment penambahan
hasil_1015 = angka1_1015 
hasil_1015 += angka2_1015
print("\nAssignment Penambahan (+=)")
print("Hasil =", hasil_1015)

#asignment pengurangan
hasil_1015 = angka1_1015
hasil_1015 -= angka2_1015
print("\nAssignment Pengurangan (-=)")
print("Hasil =", hasil_1015)

# assignment perkalian
hasil_1015 = angka1_1015
hasil_1015 *= angka2_1015
print("\nAssignment Perkalian (*=)")
print("Hasil =", hasil_1015)

# assigment pembagian, pembagian bulat, dan sisa bagi
if angka2_1015 != 0:
    hasil_1015 = angka1_1015
    hasil_1015 /= angka2_1015
    print("\nAssignment Pembagian (/=)")
    print("Hasil =", hasil_1015)
    #Operator tambahan
    hasil_1015 = angka1_1015
    hasil_1015 //= angka2_1015
    print("\nAssignment Pembagian Bulat (//=)")
    print("Hasil =", hasil_1015)

    hasil_1015 = angka1_1015
    hasil_1015 %= angka2_1015
    print("\nAssignment Sisa Bagi (%=)")
    print("Hasil =", hasil_1015)

else:
    print("Pembagian tidak dapat dilakukan")
    print("Angka kedua tidak boleh bernilai 0")

# Operator tambahan: assigment pangkat
hasil_1015 = angka1_1015
hasil_1015 **= angka2_1015
print("\nAssignment Pangkat (**=)")
print("Hasil =", hasil_1015)
```

**Penjelasan :**
1. **Penerimaan Masukan**: Menerima dua angka integer dari pengguna sebagai nilai awal `angka1_1015` dan `angka2_1015`.
2. **Operator Penugasan Biasa (`=`)**: Mengisi nilai awal ke variabel `hasil_1015`.
3. **Operator Penugasan Majemuk (*Augmented Assignment*)**: Memperbarui nilai variabel `hasil_1015` dengan mengoperasikan `angka2_1015` langsung ke dirinya sendiri menggunakan `+=`, `-=`, `*=`, `/=`, `//=`, `%=`, dan `**=`.
4. **Validasi Pembagian**: Menggunakan pengondisian `if angka2_1015 != 0` sebelum melakukan pembagian majemuk agar program tidak mengalami error *division by zero*.

### 2.2.5 Program "lainnya_2611531015.py"
Jalankan sel kode di bawah ini agar Jupyter otomatis menyimpan kodenya beserta outputnya sebagai satu kesatuan (tidak perlu screenshot terpisah untuk hasil run).

```python
# Buat file dengan nama lainnya_NIM.py
# Nama variabel ditambah 4 digit nim terakhir
# Program ini menggunakan fungsi input()
# Program operator keanggotaan dan identitas

print("===========================")
print("1. OPERATOR KEANGGOTAAN")
print("===========================")

# Input beberapa data yang dipisahkan dengan koma
input_data_1015 = input("Masukkan beberapa data (pisahkan dengan koma): ")

# Mengubah input menjadi list integer
data_1015 = [int(angka.strip()) for angka in input_data_1015.split(',')]

nilai_dicari_1015 = int(input("Masukkan angka yang ingin dicari: "))

# Operator in
hasil_1015 = nilai_dicari_1015 in data_1015
print("\nOperator keanggotaan IN")
print(nilai_dicari_1015, "in", data_1015, "=", hasil_1015)

# Operator not in  
hasil_1015 = nilai_dicari_1015 not in data_1015
print("\nOperator keanggotaan NOT IN")
print(nilai_dicari_1015, "not in", data_1015, "=", hasil_1015)

print("\n===========================")
print("2. OPERATOR IDENTITAS")
print("===========================")

# objek1 menggunakan list dari input pengguna
objek1_1015 = data_1015

# objek2 merujuk pada objek yang sama dengan objek1
objek2_1015 = objek1_1015

# objek3 memiliki isi sama, tetapi merupakan objek baru
objek3_1015 = data_1015.copy()

print("objek1 =", objek1_1015)
print("objek2 =", objek2_1015)
print("objek3 =", objek3_1015)

# Operator is
hasil_1015 = objek1_1015 is objek2_1015
print("\nOperator Identitas IS")
print("objek1 is objek2 =", hasil_1015)

# Operator is not
hasil_1015 = objek1_1015 is not objek3_1015
print("\nOperator Identitas IS NOT")
print("objek1 is not objek3 =", hasil_1015)

# Membandingkan identitas dan nilai
print("\nMembandingkan identitas dan nilai:")
print("objek1 is objek3 =", objek1_1015 is objek3_1015)
print("objek1 == objek3 =", objek1_1015 == objek3_1015)
```

**Penjelasan :**
1. **Operator Keanggotaan (*Membership*)**: Meminta masukan deretan angka lalu mengonversinya menjadi `list`. Operator `in` mengecek keberadaan nilai, sedangkan `not in` mengecek ketiadaan nilai dalam `list`.
2. **Operator Identitas (*Identity*)**: Operator `is` mengecek apakah dua variabel menunjuk ke lokasi memori yang sama (`objek1 is objek2`), sedangkan `is not` mengecek perbedaan alamat memori.
3. **Perbedaan `is` dan `==`**: Membuktikan bahwa `objek1 is objek3` bernilai `False` karena `objek3` disalin (`.copy()`) ke alamat memori baru, namun `objek1 == objek3` bernilai `True` karena isi nilainya sama persis.

### 2.2.6 Program "bitwise_2611531015.py"
Jalankan sel kode di bawah ini agar Jupyter otomatis menyimpan kodenya beserta outputnya sebagai satu kesatuan (tidak perlu screenshot terpisah untuk hasil run).

```python
# Buat file dengan nama bitwise_NIM.py
# Nama variabel ditambah 4 digit nim terakhir
# Program ini menggunakan fungsi input()

print("===========================")
print("3. OPERATOR BITWISE")
print("===========================")

angka1_1015 = int(input("Masukkan angka bitwise-1: "))
angka2_1015 = int(input("Masukkan angka bitwise-2: "))

print("\nAngka dalam bentuk desimal dan biner")
print("angka1 =", angka1_1015, "| biner =", bin(angka1_1015))
print("angka2 =", angka2_1015, "| biner =", bin(angka2_1015))

# Bitwise AND
hasil_1015 = angka1_1015 & angka2_1015
print("\nOperator Bitwise AND (&)")
print(angka1_1015, "&", angka2_1015, "=", hasil_1015)
print("Biner hasil=", bin(hasil_1015))
print("Biner hasil (8 bit) =", format(hasil_1015, '08b'))

# Bitwise OR
hasil_1015 = angka1_1015 | angka2_1015
print("\nOperator Bitwise OR (|)")
print(angka1_1015, "|", angka2_1015, "=", hasil_1015)
print("Biner hasil=", bin(hasil_1015))
print("Biner hasil (8 bit) =", format(hasil_1015, '08b'))

# Bitwise XOR
hasil_1015 = angka1_1015 ^ angka2_1015
print("\nOperator Bitwise XOR (^)")
print(angka1_1015, "^", angka2_1015, "=", hasil_1015)
print("Biner hasil=", bin(hasil_1015))
print("Biner hasil (8 bit) =", format(hasil_1015, '08b'))

# Bitwise NOT
hasil_1015 = ~angka1_1015
print("\nOperator Bitwise NOT (~)")
print("~", angka1_1015, "=", hasil_1015)
print("Biner hasil=", bin(hasil_1015))
print("Biner hasil (8 bit) =", format(hasil_1015, '08b'))

# Bitwise geser kiri
jumlah_geser_1015 = int(input("\nMasukkan jumlah pergeseran bit: "))

hasil_1015 = angka1_1015 << jumlah_geser_1015
print("\nOperator Bitwise Geser Kiri (<<)")
print(angka1_1015, "<<", jumlah_geser_1015, "=", hasil_1015)
print("Biner hasil=", bin(hasil_1015))
print("Biner hasil (8 bit) =", format(hasil_1015, '08b'))

# Bitwise geser kanan
hasil_1015 = angka1_1015 >> jumlah_geser_1015
print("\nOperator Bitwise Geser Kanan (>>)")
print(angka1_1015, ">>", jumlah_geser_1015, "=", hasil_1015)
print("Biner hasil=", bin(hasil_1015))
print("Biner hasil (8 bit) =", format(hasil_1015, '08b'))
```

**Penjelasan :**
1. **Konversi Biner**: Mengubah input nilai desimal ke biner menggunakan fungsi `bin()` dan `format(..., '08b')`.
2. **Operator Bitwise Logika**: Menjalankan operasi biner AND (`&`), OR (`|`), XOR (`^`), dan NOT (`~`).
3. **Operator Pergeseran Bit (*Bit Shift*)**: Menggeser bit ke kiri (`<<`) untuk pengalian biner dan ke kanan (`>>`) untuk pembagian bulat biner.

## **Jawaban Pertanyaan Analisis Laporan :**

### 1. Operator Aritmatika
Operator aritmatika yang digunakan dalam program beserta fungsinya adalah:
* **Perkalian (`*`):** Digunakan untuk menghitung nominal diskon berdasarkan persentase (misalnya `total_belanja * 0.10`).
* **Pengurangan (`-`):** Diaplikasikan melalui operator *augmented assignment* (`-=`) untuk memotong nilai total belanja dengan besar diskon.
* **Pembagian (`/`):** Digunakan untuk menghitung rata-rata harga barang per unit (`total_belanja / jumlah_barang`). Hasil pembagian ini menghasilkan tipe data `float`.
* **Modulo / Sisa Bagi (`%`):** Digunakan untuk mengetahui sisa jumlah barang dari pembagian kelipatan tertentu (misalnya `jumlah_barang % 3`), yang berguna untuk skema promo berbasis paket/kelipatan.

### 2. Perbedaan Operator Logika (`and`, `or`, `not`)
* **Operator `and`:** Mengembalikan nilai `True` **hanya jika kedua kondisi** di sebelah kiri dan kanannya bernilai `True`.
  * *Penerapan:* Digunakan pada `dapat_diskon` di mana pelanggan harus memenuhi status `member` **dan** total belanja `>= 200000`.
* **Operator `or`:** Mengembalikan nilai `True` jika **salah satu atau kedua kondisi** bernilai `True`.
  * *Penerapan:* Digunakan pada kelayakan promo, di mana pelanggan bisa mendapatkan promo jika `kode_promo` valid **atau** `jumlah_barang >= 3`.
* **Operator `not`:** Membalikkan nilai kebenaran (*boolean*). Kondisi `True` menjadi `False`, dan sebaliknya.
  * *Penerapan:* Digunakan untuk mengecek kondisi seperti `not (status == "nonmember")`.

### 3. Fungsi Operator Keanggotaan (`in` dan `not in`)
* **Operator `in`:** Memeriksa apakah suatu nilai terdapat di dalam suatu urutan atau koleksi data (`list`).
  * *Penerapan:* `kode_promo in daftar_promo` digunakan untuk memvalidasi apakah kode promo yang dimasukkan pengguna terdaftar dalam daftar promo toko yang valid.
* **Operator `not in`:** Memeriksa apakah suatu nilai **tidak ada** di dalam urutan atau koleksi data.
  * *Penerapan:* `kode_promo not in daftar_promo` mengembalikan `True` jika pengguna memasukkan kode promo yang tidak resmi/kadaluwarsa.

### 4. Perbedaan Operator `is` dan `==` pada Python
* **Operator `==` (Equality):** Membandingkan **nilai atau isi** dari dua objek. Mengembalikan `True` selama isi datanya sama, meskipun lokasi memorinya berbeda.
* **Operator `is` (Identity):** Membandingkan **identitas objek (alamat memori/ID RAM)** dari dua variabel. Mengembalikan `True` hanya jika kedua variabel merujuk pada objek yang sama di dalam memori (`id(a) == id(b)`).
  * *Catatan pada program:* Dua variabel string dengan nilai yang sama hasil pembuatan secara eksplisit atau dari input pengguna dapat menghasilkan `==` bernilai `True`, namun `is` bernilai `False` karena menempati alamat memori yang berbeda.

### 5. Penggunaan Operator Bitwise untuk Hak Akses
Program menggunakan bit biner (4-bit flag) untuk merepresentasikan kondisi transaksi:
* **Bitwise OR (`|`):** Digunakan untuk menggabungkan beberapa *flag* kondisi (Member, Belanja $\ge$ Rp200.000, Barang $\ge$ 3, Kode Promo) menjadi satu nilai integer `kode_status`.
* **Bitwise AND (`&`):** Digunakan untuk *masking* atau mengisolasi bit tertentu guna memeriksa apakah kondisi spesifik aktif (misalnya `kode_status & 0b0001` untuk mengecek status member).
* **Bitwise XOR (`^`):** Digunakan untuk membandingkan `kode_status` transaksi saat ini dengan `kode_referensi` standar untuk melihat perbedaan bit/akses yang belum terpenuhi.

### 6. Alur Logika Program
1. **Input Data:** Meminta input dari pengguna berupa nama, status pelanggan, total belanja, jumlah barang, dan kode promo.
2. **Validasi Kriteria:**

3. **Penyusunan Kode Status (Bitwise):** Mengompilasi status validasi ke dalam format biner 4-bit menggunakan bitwise `|` dan melakukan pengujian dengan `&`, `^`, serta shift `<<`.
4. **Kalkulasi Transaksi (Aritmatika & Penugasan):** Menghitung besar diskon jika syarat terpenuhi, memotong total belanja dengan operator `-=`, serta menghitung rata-rata harga per barang.
5. **Output:** Menampilkan ringkasan data transaksi, status validasi, rincian pembayaran, hak akses, dan detail operasi bitwise.

### 7. Identifikasi Kondisi True dan False

Berdasarkan praktikum mengenai **Operator pada Python** yang telah dilakukan, dapat ditarik beberapa kesimpulan sebagai berikut:

1. Operator Aritmatika (`+`, `-`, `*`, `/`, `//`, `%`, `**`) digunakan untuk kalkulasi matematis, sedangkan operator penugasan majemuk (seperti `+=`, `-=`, `*=`) menyederhanakan penulisan pembaruan nilai variabel.
2. Operator Perbandingan menghasilkan keluaran boolean (`True`/`False`), yang kemudian dapat digabungkan menggunakan operator logika (`and`, `or`, `not`) untuk membentuk pengondisian alur program yang kompleks.
3. Operator Keanggotaan (`in` dan `not in`) memudahkan pencarian elemen di dalam struktur data koleksi (`list`). Sementara itu, operator `is` membandingkan alamat memori objek, berbeda dengan operator `==` yang hanya membandingkan persamaan nilainya.
4. Operator Bitwise (`&`, `|`, `^`, `~`, `<<`, `>>`) memproses data langsung pada level bit biner, yang sangat berguna untuk pengolahan *flag* kondisi dan pengaturan hak akses secara efisien.
